```python
lendarios = []
miticos = []
erros = []
comuns = []
for nome in id_pokemons.values():
    try:
        dados_pokemon = requests.get(f"https://pokeapi.co/api/v2/pokemon-species/{nome}/").json()
        if dados_pokemon['is_legendary']:
            lendarios.append(dados_pokemon['name'])
        elif dados_pokemon['is_mythical']:
            miticos.append(dados_pokemon['name'])

        else:
            comuns.append(nome)
    except Exception as e:
        print(f'Erro {nome}: {e}')
        erros.append((nome,e))
        continue
    
    
```

```python
len(lendarios), len(miticos), len(comuns), len(erros)
```

```python
raridade = {
    "Lendarios" : lendarios,
    "Miticos" : miticos,
    "Comuns" : comuns
}

with open("raridade.json", "w", encoding="utf-8") as arquivo:
    json.dump(raridade, arquivo, indent=4, ensure_ascii=False)
```
